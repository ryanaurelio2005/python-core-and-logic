## 1. Estruturas de Decisão (`if`, `elif`, `else`)
Permitem executar blocos de instruções com base em condições avaliadas como verdadeiras (`True`) ou falsas (`False`).


```python
score = 78

if score >= 90:
    classificacao = "Excelente (A)"
elif score >= 70:
    classificacao = "Bom (B)"
elif score >= 50:
    classificacao = "Regular (C)"
else:
    classificacao = "Insuficiente (D)"

print(f"Pontuação: {score} -> Classificação: {classificacao}")
```

## 2. Operadores Relacionais e Lógicos
- Relacionais: `==` (igual), `!=` (diferente), `<`, `>`, `<=`, `>=`
- Lógicos: `and` (ambos verdadeiros), `or` (pelo menos um verdadeiro), `not` (inversão)


```python
idade = 20
possui_cnh = True
tem_acompanhante = False

# Condição combinada
pode_dirigir = idade >= 18 and possui_cnh
print(f"Pode dirigir legalmente: {pode_dirigir}")

# Validação alternativa com OR
pode_entrar_no_evento = idade >= 18 or tem_acompanhante
print(f"Entrada autorizada no evento: {pode_entrar_no_evento}")
```

---
## 3. Expressões Condicionais Ternárias
Permite atribuir valores a uma variável em uma única linha usando a sintaxe `valor_se_true if condicao else valor_se_false`.


```python
idade = 18
status = "Maior de idade" if idade >= 18 else "Menor de idade"
print(f"Idade {idade}: {status}")

numero = 42
paridade = "Par" if numero % 2 == 0 else "Ímpar"
print(f"O número {numero} é: {paridade}")
```

## 4. Módulo `random` e Simulações
Geração de números pseudoaleatórios para validações e dinâmicas de teste.


```python
import random

random.seed(42)  # Semente para reprodutibilidade
numero_secreto = random.randint(1, 10)
tentativa = 7

print(f"Número secreto sorteado: {numero_secreto}")
if tentativa == numero_secreto:
    print(f"Parabéns! Você acertou com o palpite {tentativa}!")
elif tentativa > numero_secreto:
    print(f"O palpite {tentativa} foi maior que o número secreto.")
else:
    print(f"O palpite {tentativa} foi menor que o número secreto.")
```

### Verificador de Ano Bissexto
Regra do calendário gregoriano: Um ano é bissexto se for divisível por 4, exceto múltiplos de 100 que não sejam múltiplos de 400.


```python
def verificar_bissexto(ano: int) -> bool:
    """Retorna True se o ano for bissexto, caso contrário False."""
    return (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0)

anos_teste = [1900, 2000, 2024, 2026]
for ano in anos_teste:
    resultado = "Bissexto" if verificar_bissexto(ano) else "Não Bissexto"
    print(f"Ano {ano:<4} -> {resultado}")
```
