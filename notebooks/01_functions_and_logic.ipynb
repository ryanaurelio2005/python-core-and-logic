## 1. Fundamentos de Funções

Funções são blocos reutilizáveis de código projetados para executar tarefas específicas.
- `def`: Palavra-chave usada para definir a assinatura da função.
- `return`: Instrução responsável por devolver o resultado processado ao chamador, permitindo atribuição direta a variáveis.


```python
# 1.1 Declaração básica de função com tipagem (Type Hints) e retorno
def somar(a: float, b: float) -> float:
    """Calcula e retorna a soma de dois números."""
    return a + b

# Execução e atribuição do resultado
resultado = somar(2, 3)
print(f"Resultado da soma: {resultado}")
```

```python
# 1.2 Formatação e interpolação de strings em funções
def formatar_mensagem(nome: str, sufixo: str = "<--- referência") -> str:
    """Retorna uma mensagem formatada utilizando o argumento recebido."""
    return f"Valor processado: {nome} {sufixo}"

msg = formatar_mensagem("Atualização de estado concluída com sucesso")
print(msg)
```

## 2. Parâmetros Padrão (*Default*) e Obrigatórios

Em Python, parâmetros com valores predefinidos (*default arguments*) tornam a chamada da função mais flexível.
> **Regra importante:** Parâmetros obrigatórios (*non-default*) devem sempre vir antes dos parâmetros opcionais (*default*).


```python
def exibir_perfil(nome: str, idade: int, cidade: str = "Goiânia") -> None:
    """Exibe informações de perfil do usuário, utilizando cidade padrão quando omitida."""
    print(f"Nome: {nome:<10} | Idade: {idade:>2} anos | Cidade: {cidade}")

# Chamada fornecendo todos os parâmetros
exibir_perfil("Ryan", 21, "Goiânia")

# Chamada omitindo o parâmetro default 'cidade'
exibir_perfil("Leah", 22)
```

## 3. Argumentos Dinâmicos com `*args`

O operador `*args` permite que uma função receba uma quantidade arbitrária de argumentos posicionais. Internamente, os valores são empacotados em uma **tupla imutável**.


```python
# 3.1 Somatório com número variável de argumentos
def somar_todos(*valores: float) -> float:
    """Calcula o somatório de N argumentos passados dinamicamente."""
    return sum(valores)

total = somar_todos(1, 2, 3, 4, 5, 6, 7)
print(f"Somatório total de argumentos: {total}")
```

```python
# 3.2 Estatísticas descritivas (Média, Máximo e Mínimo)
from typing import Tuple

def calcular_estatisticas(*valores: float) -> Tuple[float, float, float]:
    """Calcula a média aritmética, o maior e o menor valor de uma sequência numérica.
    
    Raises:
        ValueError: Caso nenhum valor seja passado para a função.
    """
    if not valores:
        raise ValueError("A coleção de valores não pode ser vazia.")
    
    media = sum(valores) / len(valores)
    maior = max(valores)
    menor = min(valores)
    
    return media, maior, menor

# Execução com múltiplos números
media, maior, menor = calcular_estatisticas(10, 25, 42, 8, 95, 14)
print(f"Média: {media:.2f} | Maior: {maior} | Menor: {menor}")
```

## 4. Filtragem e Validação de Dados

Boas práticas de desenvolvimento exigem:
1. Validação defensiva com `try...except ValueError` para entradas do usuário.
2. Uso de **List Comprehensions** para código conciso, performático e legível.


```python
# 4.1 Separação e agregação condicional de valores numéricos
def analisar_conjunto(numeros: list[int]) -> dict:
    """Classifica uma lista de números em positivos, negativos, zeros e pares."""
    positivos = [n for n in numeros if n > 0]
    negativos = [n for n in numeros if n < 0]
    zeros = [n for n in numeros if n == 0]
    pares = [n for n in numeros if n % 2 == 0]
    
    return {
        "soma_positivos": sum(positivos),
        "soma_negativos": sum(negativos),
        "total_zeros": len(zeros),
        "total_pares": len(pares),
        "total_elementos": len(numeros)
    }

# Amostra de teste contendo números variados
dados_amostra = [-5, -3, -1, 0, 0, 2, 4, 6, 8, 10]
relatorio_dados = analisar_conjunto(dados_amostra)

print("Relatório de Análise Numérica:")
print("-" * 35)
for chave, valor in relatorio_dados.items():
    nome_formatado = chave.replace('_', ' ').capitalize()
    print(f"{nome_formatado:<20}: {valor}")
```

```python
# 4.2 Pipeline seguro para tratamento de entradas textuais
def processar_entrada_texto(entrada_raw: str) -> None:
    """Converte texto separado por espaços em números e calcula métricas completas."""
    entrada = entrada_raw.strip()
    if not entrada:
        print("Erro: Nenhuma entrada foi fornecida.")
        return

    try:
        valores = [int(item) for item in entrada.split()]
        media, maior, menor = calcular_estatisticas(*valores)
        pares = [n for n in valores if n % 2 == 0]

        print(f"Valores informados : {valores}")
        print(f"Soma total         : {sum(valores)}")
        print(f"Média calculada    : {media:.2f}")
        print(f"Maior valor        : {maior}")
        print(f"Menor valor        : {menor}")
        print(f"Total de pares     : {len(pares)}")
    except ValueError:
        print("Erro de validação: Insira apenas números inteiros separados por espaço.")

# Demonstração do pipeline
processar_entrada_texto("12 5 8 20 3 15 4")
```

## 5. Estudo de Caso 1: Monitoramento de Temperatura

Sistema para suporte meteorológico e agronômico que avalia intervalos de temperatura, conforto térmico e amplitude térmica.


```python
def classificar_temperatura(temp: float) -> str:
    """Classifica a temperatura em Frio (< 18°C), Ideal (18°C a 30°C) ou Quente (> 30°C)."""
    if temp < 18.0:
        return "Frio"
    elif 18.0 <= temp <= 30.0:
        return "Ideal"
    return "Quente"

def calcular_amplitude(t_min: float, t_max: float) -> float:
    """Calcula a amplitude térmica absoluta."""
    return abs(t_max - t_min)

def emitir_boletim_climatico(temp_min: float, temp_max: float) -> None:
    """Exibe relatório meteorológico formatado."""
    amp = calcular_amplitude(temp_min, temp_max)
    media = (temp_min + temp_max) / 2
    
    print("=" * 40)
    print("        BOLETIM METEOROLÓGICO        ")
    print("=" * 40)
    print(f"Temperatura Mínima : {temp_min:>5.1f}°C  ({classificar_temperatura(temp_min)})")
    print(f"Temperatura Máxima : {temp_max:>5.1f}°C  ({classificar_temperatura(temp_max)})")
    print(f"Amplitude Térmica  : {amp:>5.1f}°C")
    print(f"Média Diária       : {media:>5.1f}°C")
    print("=" * 40)

# Simulação com temperaturas registradas
emitir_boletim_climatico(temp_min=23.2, temp_max=25.0)
```

## 6. Estudo de Caso 2: Análise de Produtividade Agrícola

Módulo de gestão agrícola que monitora o rendimento de múltiplos talhões de cultivo, gera estatísticas consolidadas da safra e aponta os talhões de maior destaque com tratamento para empates.


```python
def classificar_produtividade(rendimento: float) -> str:
    """Classifica a produtividade do talhão com base no volume colhido."""
    if rendimento <= 50.0:
        return "Baixa produtividade"
    elif 51.0 <= rendimento <= 80.0:
        return "Boa produtividade"
    return "Excelente produtividade"

def gerar_relatorio_safra(*talhoes: float) -> None:
    """Gera um relatório gerencial consolidado para N talhões da fazenda."""
    if not talhoes:
        print("Aviso: Nenhum talhão informado.")
        return

    media_safra = sum(talhoes) / len(talhoes)
    maior_rendimento = max(talhoes)
    
    # Identifica todos os talhões que atingiram a produtividade máxima (trata empates)
    destaques = [f"Talhão {i+1}" for i, val in enumerate(talhoes) if val == maior_rendimento]

    print("=" * 48)
    print("       RELATÓRIO DE PRODUTIVIDADE AGRÍCOLA       ")
    print("=" * 48)
    for i, rendimento in enumerate(talhoes, start=1):
        status = classificar_produtividade(rendimento)
        print(f"Talhão {i:02d}: {rendimento:>5.1f} sc/ha | Classificação: {status}")
    
    print("-" * 48)
    print(f"Média Geral da Safra : {media_safra:.1f} sc/ha")
    print(f"Maior Produtividade  : {maior_rendimento:.1f} sc/ha ({', '.join(destaques)})")
    print("=" * 48)

# Simulação com 3 talhões (incluindo caso de empate)
gerar_relatorio_safra(80.0, 80.0, 60.0)
```
