## 1. Anatomia de Funções
Funções encapsulam blocos lógicos reutilizáveis. O uso de *Type Hints* documenta a tipagem de entrada e saída.


```python
def somar(a: float, b: float) -> float:
    """Calcula e retorna a soma de dois números."""
    return a + b

resultado = somar(12.5, 7.5)
print(f"Resultado da soma: {resultado}")
```

---
## 2. Parâmetros Dinâmicos (`*args` e `**kwargs`)
- `*args`: Empacota múltiplos parâmetros posicionais em uma **tupla**.
- `**kwargs`: Empacota parâmetros nomeados em um **dicionário**.


```python
def registrar_sensor(**kwargs) -> dict:
    """Recebe parâmetros nomeados flexíveis para cadastro de sensores."""
    return kwargs

sensor_info = registrar_sensor(
    id="SENSOR-AGRO-01",
    tipo="Umidade de Solo",
    setor="Talhão 03",
    frequencia_minutos=15
)

print("Dados do sensor registrado:")
for chave, valor in sensor_info.items():
    print(f"{chave:<20}: {valor}")
```

```python
from typing import Tuple

def calcular_estatisticas(*valores: float) -> Tuple[float, float, float]:
    """Calcula média aritmética, valor máximo e valor mínimo."""
    if not valores:
        raise ValueError("A coleção de valores não pode ser vazia.")
    return sum(valores) / len(valores), max(valores), min(valores)

media, maior, menor = calcular_estatisticas(15, 28, 42, 9, 33)
print(f"Média: {media:.2f} | Maior: {maior} | Menor: {menor}")
```

## 3. Escopos e Closures
Controle de escopo local, global e geração dinâmica de funções (*Factory Functions* com `nonlocal`).


```python
def criar_multiplicador(fator: float):
    """Factory function que retorna uma nova função multiplicadora (Closure)."""
    return lambda x: x * fator

dobro = criar_multiplicador(2)
triplo = criar_multiplicador(3)

print(f"Dobro de 15: {dobro(15)}")
print(f"Triplo de 15: {triplo(15)}")
```

## 4. Lambdas e Higher-Order Functions (`map`, `filter`)
Expressões anônimas concisas para pipelines de dados funcionais.


```python
temperaturas_celsius = [18, 22, 25, 28, 32, 36]

# Conversão de °C para °F usando map
c_para_f = lambda c: (c * 9/5) + 32
temperaturas_fahrenheit = list(map(c_para_f, temperaturas_celsius))

# Filtragem de temperaturas altas (>= 25°C) usando filter
temperaturas_quentes = list(filter(lambda c: c >= 25, temperaturas_celsius))

print(f"Temperaturas em Celsius    : {temperaturas_celsius}")
print(f"Temperaturas em Fahrenheit : {temperaturas_fahrenheit}")
print(f"Temperaturas elevadas (>=25): {temperaturas_quentes}")
```

## 5. Estudos de Caso

### Monitoramento de Umidade do Solo


```python
def classificar_umidade_solo(umidade: float) -> str:
    """Classifica a condição hídrica do solo para tomada de decisão de irrigação."""
    if umidade < 30.0:
        return "Solo Seco -> Acionar Irrigação"
    elif 30.0 <= umidade <= 60.0:
        return "Umidade Ideal -> Manter monitoramento"
    return "Solo Saturado -> Suspender Irrigação"

leituras_solo = [22.5, 45.0, 72.0]
for u in leituras_solo:
    print(f"Umidade: {u:>5.1f}% | Diagnóstico: {classificar_umidade_solo(u)}")
```

### Gestão de Produtividade Agrícola (Talhões)

```python
def classificar_produtividade(rendimento: float) -> str:
    if rendimento <= 50.0:
        return "Baixa produtividade"
    elif 51.0 <= rendimento <= 80.0:
        return "Boa produtividade"
    return "Excelente produtividade"

def gerar_relatorio_safra(*talhoes: float) -> None:
    if not talhoes:
        print("Aviso: Nenhum talhão informado.")
        return

    media_safra = sum(talhoes) / len(talhoes)
    maior_rendimento = max(talhoes)
    destaques = [f"Talhão {i+1}" for i, val in enumerate(talhoes) if val == maior_rendimento]

    print("=" * 48)
    print("       RELATÓRIO DE PRODUTIVIDADE AGRÍCOLA       ")
    print("=" * 48)
    for i, rendimento in enumerate(talhoes, start=1):
        status = classificar_produtividade(rendimento)
        print(f"Talhão {i:02d}: {rendimento:>5.1f} sc/ha | Classificação: {status}")
    
    print("-" * 48)
    print(f"Média Geral da Safra : {media_safra:.1f} sc/ha")
    print(f"Maior Produtividade  : {maior_rendimento:.1f} sc/ha ({', '.join(destaques)})")
    print("=" * 48)

gerar_relatorio_safra(80.0, 80.0, 60.0)
```
