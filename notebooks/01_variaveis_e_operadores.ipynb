## 1. Tipos Primitivos
Python possui tipagem dinâmica e forte. Os quatro tipos fundamentais são:
- `int`: Números inteiros (ex: 10, -5, 0)
- `float`: Números decimais / ponto flutuante (ex: 3.14, 99.90)
- `str`: Textos e cadeias de caracteres (ex: "Python", 'Ryan')
- `bool`: Valores lógicos booleanos (`True` ou `False`)


```python
# Declaração de variáveis primitivas
nome = "Ryan"
idade = 21
saldo = 1250.75
ativo = True

print(f"Nome : {nome:<10} | Tipo: {type(nome).__name__}")
print(f"Idade: {idade:<10} | Tipo: {type(idade).__name__}")
print(f"Saldo: R$ {saldo:<7.2f} | Tipo: {type(saldo).__name__}")
print(f"Ativo: {str(ativo):<10} | Tipo: {type(ativo).__name__}")
```

## 2. Operações Aritméticas
Python suporta todas as operações matemáticas básicas com suporte nativo a exponenciação e divisão inteira:


```python
a = 15
b = 4

print(f"Soma (+)             : {a} + {b} = {a + b}")
print(f"Subtração (-)        : {a} - {b} = {a - b}")
print(f"Multiplicação (*)    : {a} * {b} = {a * b}")
print(f"Divisão Real (/)     : {a} / {b} = {a / b:.2f}")
print(f"Divisão Inteira (//) : {a} // {b} = {a // b}")
print(f"Módulo / Resto (%)   : {a} % {b} = {a % b}")
print(f"Exponenciação (**)   : {a} ** {b} = {a ** b}")
```

## 3. Entrada, Saída e Interpolação de Strings
A função `print()` com f-strings permite formatações precisas de casas decimais e alinhamento.


```python
produto = "Notebook Dell"
preco_unitario = 4599.90
quantidade = 2
total = preco_unitario * quantidade

# Interpolação elegante com alinhamento e formatação financeira
print("=" * 45)
print(f"Produto    : {produto}")
print(f"Preço Unit.: R$ {preco_unitario:>10.2f}")
print(f"Quantidade : {quantidade:>10}")
print(f"Valor Total: R$ {total:>10.2f}")
print("=" * 45)
```

## 4. Conversão de Tipos (*Type Casting*)
Dados lidos via `input()` chegam sempre como texto (`str`), exigindo conversão explícita para `int` ou `float`.


```python
# Simulação de parsing de string para tipos numéricos
entrada_idade = "25"
entrada_peso = "78.5"

idade_convertida = int(entrada_idade)
peso_convertido = float(entrada_peso)

print(f"Idade em dobro: {idade_convertida * 2}")
print(f"Peso arredondado: {round(peso_convertido)} kg")
```

### Exercício 1: Cálculo de Média Escolar


```python
nota1 = 7.5
nota2 = 8.3
nota3 = 6.9

media = (nota1 + nota2 + nota3) / 3
print(f"Notas registradas: {nota1}, {nota2}, {nota3}")
print(f"Média Final: {media:.2f}")
```

### Exercício 2: Cálculo de Idade com Base no Ano de Nascimento

```python
ano_nascimento = 2005
ano_atual = 2026

idade_calculada = ano_atual - ano_nascimento
print(f"Nascido em: {ano_nascimento}")
print(f"Idade estimada em {ano_atual}: {idade_calculada} anos")
```
