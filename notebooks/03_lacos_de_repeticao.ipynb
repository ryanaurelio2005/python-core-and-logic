## 1. Laço `for` e Função `range()`
O comando `for` itera sobre sequências. O gerador `range(start, stop, step)` produz intervalos numéricos eficientes.


```python
# Tabuada automatizada do 7
numero_base = 7
print(f"=== TABUADA DO {numero_base} ===")
for i in range(1, 11):
    print(f"{numero_base} x {i:>2} = {numero_base * i:>2}")
```

## 2. Acumuladores, Contadores e Fatorial
Técnica para agregação de valores ao longo de iterações sucessivas.


```python
# Cálculo de somatório acumulado
limite = 10
soma_acumulada = 0

for i in range(1, limite + 1):
    soma_acumulada += i

print(f"Soma dos números de 1 até {limite}: {soma_acumulada}")

# Cálculo de Fatorial
n = 5
fatorial = 1
for i in range(1, n + 1):
    fatorial *= i

print(f"Fatorial de {n}! = {fatorial}")
```

## 3. Laços Aninhados (*Nested Loops*) e Padrões
Combinação de loops externos e internos para navegação em matrizes ou criação de formatos visuais.


```python
# Construção de pirâmide simétrica de caracteres
altura = 5

for i in range(altura):
    espacos = " " * (altura - i - 1)
    estrelas = "*" * (2 * i + 1)
    print(f"{espacos}{estrelas}")
```

## 4. Laço `while`
Repete um bloco de comandos enquanto uma condição lógica for verdadeira.


```python
contador = 1
print("Contagem progressiva com while:")
while contador <= 5:
    print(f"Iteração número {contador}")
    contador += 1
```

---
## 5. Controle de Fluxo com `break` e `continue`
- `break`: Interrompe a execução do laço imediatamente.
- `continue`: Pula para a próxima iteração do laço, ignorando o código restante no ciclo atual.


```python
# Demonstração de continue (pulando ímpares) e break (interrompendo em 8)
print("Números pares até encontrar 8:")
for num in range(1, 15):
    if num % 2 != 0:
        continue  # Pula os ímpares
    if num > 8:
        break     # Encerra o laço
    print(f"Valor aceito: {num}")
```

## 6. Projeto Prático: Simulador de Caixa Eletrônico
Simulação com máquina de estados e validações de saldo.


```python
class CaixaEletronico:
    def __init__(self, saldo_inicial: float = 1000.0):
        self.saldo = saldo_inicial

    def depositar(self, valor: float) -> str:
        if valor <= 0:
            return "Valor inválido para depósito."
        self.saldo += valor
        return f"Depósito de R$ {valor:.2f} realizado com sucesso."

    def sacar(self, valor: float) -> str:
        if valor <= 0:
            return "Valor inválido para saque."
        if valor > self.saldo:
            return f"Saldo insuficiente. Saldo disponível: R$ {self.saldo:.2f}."
        self.saldo -= valor
        return f"Saque de R$ {valor:.2f} efetuado com sucesso."

caixa = CaixaEletronico(saldo_inicial=500.0)
print(caixa.depositar(250.0))
print(caixa.sacar(100.0))
print(caixa.sacar(800.0))
print(f"Saldo final em conta: R$ {caixa.saldo:.2f}")
```
